# Practical 4: Gradient Descent Algorithm

**Aim:** Find the local minimum of `y = (x + 3)^2` starting from `x = 2`.

**Logic:** Gradient Descent updates x using:

`new_x = x - learning_rate * derivative`

For this function, the derivative is `2 * (x + 3)`. The minimum is at `x = -3`, where `y = 0`.

Run each cell using `Shift + Enter`.

## Cell 1: Import Library

```python
import matplotlib.pyplot as plt
```

## Cell 2: Define the Function and Derivative

```python
def f(x):
    return (x + 3) ** 2

def gradient(x):
    return 2 * (x + 3)
```

## Cell 3: Apply Gradient Descent
Start at x = 2. The learning rate controls how big each step is.

```python
x = 2
learning_rate = 0.1

for i in range(100):
    x = x - learning_rate * gradient(x)

print("Minimum x:", x)
print("Minimum y:", f(x))
```

## Cell 4: Plot the Function and Minimum

```python
x_values = [i / 10 for i in range(-50, 31)]
y_values = [f(i) for i in x_values]

plt.plot(x_values, y_values)
plt.scatter(x, f(x), color="red", label="Minimum point")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent")
plt.legend()
plt.grid()
plt.show()
```

## Result

The Gradient Descent Algorithm successfully finds the minimum of `y = (x + 3)^2`, starting from `x = 2`.

The value of x approaches **-3**, and the minimum value of y approaches **0**.
